```python
import torch
from torch.utils.data import TensorDataset, DataLoader
```

```python
# 定义数据
x = torch.randn(10, 3)
y = torch.randn(10)
print(x)
print(y)
```

```python
# 构建数据集
dataset = TensorDataset(x, y)
```

```python
# 定义加载器
dataloader = DataLoader(dataset, batch_size=4, shuffle=False, drop_last=True)
```

```python
# 遍历每个批次的数据
for x, y in dataloader:
    print(x)
    print("y = ", y)
    print()
```
